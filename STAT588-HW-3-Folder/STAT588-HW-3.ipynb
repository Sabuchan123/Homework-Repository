```python
import torch
from torchvision import datasets, transforms
mnist_train = datasets.MNIST(root='./data', train=True, download=True, transform=transforms.ToTensor())
mnist_test = datasets.MNIST(root='./data', train=False, download=True, transform=transforms.ToTensor())
print(mnist_train.data.shape)
print(mnist_test.data.shape)


all_data = torch.cat((mnist_train.data, mnist_test.data), dim=0).float()
all_targets = torch.cat((mnist_train.targets, mnist_test.targets), dim=0)

indices = torch.randperm(all_data.size(0))[:5000]
x = all_data[indices]
x = x.view(5000, 784)
x = (x - x.mean()) / x.std()

y = torch.full((5000, 10), -0.9, device='cuda')
y.scatter_(1, all_targets[indices].unsqueeze(1).to('cuda'), 0.9)

print(x.shape)
print(y.shape)

train_input = x[:4000].reshape(4000, -1).to('cuda')
train_target = y[:4000].to('cuda')
test_input = x[4000:].reshape(1000, -1).to('cuda')
test_target = y[4000:].to('cuda')

print("Train Input Shape:", train_input.shape)
print("Train Target Shape:", train_target.shape)
print("Test Input Shape:", test_input.shape)
print("Test Target Shape:", test_target.shape)
```

```python
w1 = torch.eye(784, device='cuda', dtype=torch.float32)
random_matrix = torch.normal(0, 1e-6, size=(784, 784), device='cuda', dtype=torch.float32)
w1.add_(random_matrix)
b1 = torch.normal(0, 1e-6, size=(784,1), device='cuda', dtype=torch.float32)
b2 = torch.normal(0, 1e-6, size=(784,1), device='cuda', dtype=torch.float32)
w3 = torch.normal(0, 1e-6, size=(10, 784), device='cuda', dtype=torch.float32)
b3 = torch.normal(0, 1e-6, size=(10,1), device='cuda', dtype=torch.float32)

dl_dw1 = torch.zeros_like(w1)
dl_db1 = torch.zeros_like(b1)
dl_db2 = torch.zeros_like(b2)
dl_dw3 = torch.zeros_like(w3)
dl_db3 = torch.zeros_like(b3)

def sigma(x):
    return torch.tanh(x)

def dsigma(x):
    return 1 - torch.tanh(x)**2

# test = torch.ones(5)
# print(sigma(test))
# print(dsigma(test))

def loss(v,t):
    return ((t-v)**2).sum()

def dloss(v,t):
    return 2*(v-t)

def forward_pass(w1, b1, b2, w3, b3, x):
    x0 = x.view(-1, 1) # 784 x 1
    s1 = w1 @ x0 + b1  # 784 x 1
    x1 = sigma(s1)     # 784 x 1
    s2 = w1 @ x1 + b2  # 784 x 1
    x2 = sigma(s2)     # 784 x 1
    s3 = w3 @ x2 + b3  # 10 x 1
    x3 = sigma(s3)     # 10 x 1
    return s1, x1, s2, x2, s3, x3

def backward_pass(w1, w3, t, x, s1, x1, s2, x2, s3, x3, dl_dw1, dl_db1, dl_db2, dl_dw3, dl_db3):
    x0 = x # 784 x 1
    dl_dx3 = dloss(x3, t)        # 10 x 1
    dl_ds3 = dl_dx3 * dsigma(s3) # 10 x 1 (note dsigma(s3) mathematically has a size of 10x10, but the off diagonals are all zeroes).
    dl_dx2 = w3.T @ dl_ds3       # 784 x 1 (note in the equation in problem 2, w3.T is multiplied to the right of dl_ds3, but that would work only if dl_ds3 was 10x10 and not 10x1, or if dl_ds3 was transposed)
    dl_ds2 = dl_dx2 * dsigma(s2) # 784 x 1 (note same idea applies as dl_ds3)
    dl_dx1 = w1.T @ dl_ds2       # 784 x 1 (note same as before)
    dl_ds1 = dl_dx1 * dsigma(s1) # 784 x 1 (note same as before)

    dl_dw3.add_(dl_ds3.view(-1, 1) @ x2.view(1, -1)) # 10 x 784 (note without .view, both dl_ds3 and x2 would be seen as 1d tensors)
    dl_db3.add_(dl_ds3)
    dl_dw1.add_(dl_ds2.view(-1, 1) @ x1.view(1, -1) + dl_ds1.view(-1, 1) @ x0.view(1, -1)) # 784 x 784
    dl_db2.add_(dl_ds2)
    dl_db1.add_(dl_ds1)

```

```python
eta = 0.001 / train_input.size(0)
for step in range(250):
    print("Step:", step + 1)
    dl_dw1.zero_()
    dl_db1.zero_()
    dl_db2.zero_()
    dl_dw3.zero_()
    dl_db3.zero_()

    train_loss = 0.0
    train_errors = 0
    for i in range(train_input.size(0)):
        x = train_input[i].view(-1, 1)
        t = train_target[i].view(-1, 1)
        s1, x1, s2, x2, s3, x3 = forward_pass(w1, b1, b2, w3, b3, x)
        train_errors += x3.argmax().item() != t.argmax().item()
        train_loss += loss(x3, t).item()

        backward_pass(w1, w3, t, x, s1, x1, s2, x2, s3, x3, dl_dw1, dl_db1, dl_db2, dl_dw3, dl_db3)
    print("Train Loss:", train_loss)
    print("Train Errors:", train_errors / train_input.size(0))
        
    with torch.no_grad():
        w1 -= eta * dl_dw1
        b1 -= eta * dl_db1
        b2 -= eta * dl_db2
        w3 -= eta * dl_dw3
        b3 -= eta * dl_db3

    test_errors = 0
    for i in range(test_input.size(0)):
        _, _, _, _, _, x3 = forward_pass(w1, b1, b2, w3, b3, test_input[i].view(-1, 1))
        test_errors += x3.argmax().item() != test_target[i].argmax().item()
    print("Test Errors:", test_errors / test_input.size(0))
```
